<h1> Task 1 </h1>

This first section reads the provided data sets and isolates the relevant datapoints into lists

```c++
import os
import numpy as np
import pandas as pd
import networkx as nx
import matplotlib
matplotlib.use("Agg")
import matplotlib.pyplot as plt

# files
BACI_FILE = "BACI_HS22_Y2024_V202601.csv"
COUNTRY_FILE = "country_codes_V202601.csv"
PRODUCT_FILE = "product_codes_HS22_V202601.csv"

# product code for bamboo pulp
PRODUCT_CODES = ["470630"]
YEAR = 2024
OUT_DIR = f"output_{'_'.join(PRODUCT_CODES)}"

os.makedirs(OUT_DIR, exist_ok=True)


# loads data
def load_flows():

    # loads BACI file
    chunks = pd.read_csv(
        BACI_FILE,
        dtype = {"k": str},
        na_values = ["", "NA", "NaN"],
        chunksize = 1000000,
    )

    keep = []

    # finds the bamboo pulp data from the BACI file
    for c in chunks:

        c["k"] = c["k"].str.zfill(6)

        # k is the product code and t is the year the data is from
        keep.append(c[c["k"].isin(PRODUCT_CODES) & (c["t"] == YEAR)])

    dataSet = pd.concat(keep, ignore_index=True)

    return dataSet


# loads data into lists
raw = load_flows()
countries = pd.read_csv(COUNTRY_FILE)
products = pd.read_csv(PRODUCT_FILE, dtype={"code": str})

# fixes turkey's name
def fix_mojibake(s):
    # undo "UTF-8 read as Latin-1" damage; leave normal text untouched
    try:
        return s.encode("latin-1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return s


countries["country_name"] = countries["country_name"].apply(fix_mojibake)
```

Formatts the data into a more usable format

```c++
countries["country_name"] = countries["country_name"].apply(fix_mojibake)

# organises trades into one link per trade
edges = (raw.groupby(["i", "j"], as_index=False).agg(value=("v", "sum"), volume=("q", "sum")))

# marks null values
edges.loc[edges["volume"] <= 0, "volume"] = np.nan

# data is in 1k dollars so this converts it back into single dollar value
edges["unit_value"] = edges["value"] * 1000 / edges["volume"]

# replaces country codes with names
cmap = countries.set_index("country_code")

edges["exporter"] = edges["i"].map(cmap["country_name"])
edges["importer"] = edges["j"].map(cmap["country_name"])
edges["exp_iso3"] = edges["i"].map(cmap["country_iso3"])
edges["imp_iso3"] = edges["j"].map(cmap["country_iso3"])

# saves edges as csv
edges.to_csv(f"{OUT_DIR}/edges.csv", index=False)
```

Analyses the data set and summarises relevant information

```c++
# draws network with info from edges
G = nx.DiGraph()
for r in edges.itertuples():
    G.add_edge(r.exporter, r.importer, value = r.value, volume = r.volume,unit_value = r.unit_value)

# sets distance to represent strength of trade
for u, v, d in G.edges(data = True):
    d["distance"] = 1.0 / d["value"]

N, M = G.number_of_nodes(), G.number_of_edges()


# measures stats
nodes = pd.DataFrame(index = sorted(G.nodes()))

nodes["in_degree"] = pd.Series(dict(G.in_degree()))
nodes["out_degree"] = pd.Series(dict(G.out_degree()))

nodes["import_value"] = pd.Series(dict(G.in_degree(weight = "value")))
nodes["export_value"] = pd.Series(dict(G.out_degree(weight = "value")))

nodes["import_volume"] = pd.Series(dict(G.in_degree(weight = "volume")))
nodes["export_volume"] = pd.Series(dict(G.out_degree(weight = "volume")))

nodes["net_exports_value"] = nodes["export_value"] - nodes["import_value"]
nodes["total_value"] = nodes["export_value"] + nodes["import_value"]
nodes["export_unit_value"] = nodes["export_value"] * 1000 / nodes["export_volume"].replace(0, np.nan)

# network metrics
nodes["pagerank_value"] = pd.Series(nx.pagerank(G, weight = "value"))
nodes["betweenness"] = pd.Series(nx.betweenness_centrality(G, weight = "distance"))
nodes["clustering_w"] = pd.Series(nx.clustering(G, weight = "value"))
nodes["hub_score"], nodes["authority_score"] = [pd.Series(s) for s in nx.hits(G)]

nodes.index.name = "country"

# saves to csv
nodes.to_csv(f"{OUT_DIR}/node_metrics.csv")

# Network-level metrics
UG = G.to_undirected()
weaklyConneced = list(nx.weakly_connected_components(G))
stronglyConnected = list(nx.strongly_connected_components(G))
totalValue = edges["value"].sum()

# value of how equal export levels are
# closer to 1 means only a few countries are doing the bulk
def gini(x):

    # sorts amount
    x = np.sort(np.asarray(x, dtype = float))
    num = len(x)

    ranks = (2 * np.arange(1, num + 1) - num - 1)

    value = ranks.dot(x) / (num * x.sum())
    return value

# gets top importer/ exporter
def topShare(s, k):
    return s.sort_values(ascending = False).head(k).sum() / s.sum()

summary = {
    "Countries": N,
    "Trade links": M,
    "Total trade value in millions": round(totalValue / 1e3, 1),
    "Total trade volume in thousand tons": round(edges["volume"].sum() / 1e3, 1),
    "Density": round(nx.density(G), 4),
    "Reciprocity": round(nx.reciprocity(G), 4),
    "Weakly connected components": len(weaklyConneced),
    "Largest weak component (nodes)": max(len(c) for c in weaklyConneced),
    "Strongly connected components": len(stronglyConnected),
    "Largest strong component (nodes)": max(len(c) for c in stronglyConnected),
    "Average clustering (undirected)": round(nx.average_clustering(UG), 4),
    "Average weighted clustering (directed)": round(nodes["clustering_w"].mean(), 4),
    "Transitivity (undirected)": round(nx.transitivity(UG), 4),
    "Degree assortativity (out->in)": round(nx.degree_assortativity_coefficient(G, x="out", y="in"), 4),
    "Mean in/out degree": round(M / N, 2),
    "Max out-degree (country)": f"{nodes['out_degree'].max()} ({nodes['out_degree'].idxmax()})",
    "Max in-degree (country)": f"{nodes['in_degree'].max()} ({nodes['in_degree'].idxmax()})",
    "Top-5 exporters' share of export value": round(topShare(nodes["export_value"], 5), 3),
    "Top-5 importers' share of import value": round(topShare(nodes["import_value"], 5), 3),
    "Gini of export value": round(gini(nodes["export_value"]), 3),
    "Gini of import value": round(gini(nodes["import_value"]), 3),
}


core = UG.subgraph(max(weaklyConneced, key=len))
summary["Avg shortest path (largest component, undirected)"] = round(nx.average_shortest_path_length(core), 3)
summary["Diameter (largest component, undirected)"] = nx.diameter(core)

# saves above info into summary.txt
with open(f"{OUT_DIR}/summary.txt", "w") as f:
    for k, v in summary.items():
        f.write(f"{k}: {v}\n")
        print(f"{k}: {v}")
```

<h2> Question 2 </h2>

The national population data was taken from the [world bank population records](https://data.worldbank.org/indicator/SP.POP.TOTL). This questions asks us to find which countries are more "important" than would be expected from their population size. I chose to interpret "importance" by the value of trade they export every year. Since bamboo pulp export is mainly dominated by 10 countires, where imports are more evenly dispursed, the importance of a country in terms of bamboo pulp trade is determined by how much they participate in its export.

```c++
# drops countries with only imports
sub = df[df[m] > 0]

# normalises population and trade values
x, y = np.log10(sub["population"]), np.log10(sub[m])

# finds slope
slope, icpt = np.polyfit(x, y, 1)

# difference between a countries trade value and what the slope predits it should have based on its population
resid = y - (slope * x + icpt)

# amount of relation between pop and trade value
r2 = 1 - (resid.var() / y.var())
z = resid / resid.std()
```

The code above describes the method used to determine and plot the relation between population and trade value. It first excludes any countries that dont export the goods. It then uses a double log method to normalise both the population and trade value, since in this case both values are heavily skewed (trade value specifically since 2 countries do a majority of the exporting). The slope is then estimated so the model learns from the data how trade normally scales with population, rather than assuming it scales proportionally. It then measures the distance between the actual trade value and the norm laid out by the slope.
